```python
# filtering using pandas : keeping the rows that match conditions
df = pd.read_csv("Friends.csv",index_col="Name") # ascending rows by Name 

df = df.iloc[:,0:4]
print(df)
old_friend = df[df["Age"]>= 29]
print(old_friend)
young_friend = df[df["Age"]<=28]
print(young_friend)
married_young_friend = df[(df["IsMarried"] ==True ) | (df["Age"] <=29 )  ]
print(married_young_friend)
unmarried_old_firend = df[(df["IsMarried"] ==False ) & (df["Age"] >30 )  ]
print(unmarried_old_firend)

```

```python
# aggregate functions - Reduces a set of values into single summary values
# used to summarize and analyze data and often used with the groupby() functions
# this function will apply to whole data frame
df = pd.read_csv("Friends.csv",index_col="Name") # ascending rows by Name 
print(df)
print("-------------------------")
print(df.mean(numeric_only=True)) # average
print(df.sum(numeric_only=True)) # sum
print(df.min(numeric_only=True)) # min
print(df.max(numeric_only=True)) # max
print(df.count()) # count
```

```python
# single column aggregate function
df = pd.read_csv("friends_new.csv",index_col="Name") # ascending rows by Name 
# print(df)
print("-------------------------")
print(df["Age"].mean()) # average
print(df["Age"].sum()) # sum
print(df["Age"].min()) # min
print(df["Age"].max()) # max
print(df["Age"].count()) # count
```

```python
# group by function
df = pd.read_csv("friends_new.csv",index_col="Name") # ascending rows by Name 
# print(df)
# select department_name,avg(salary) from empl group by department_name
print("------------mean-------------")
group_df = df.groupby("Job")
print(group_df["Age"].mean())
print("--------------sum-----------")
print(group_df["Age"].sum())
print("----------------min---------")
print(group_df["Age"].min())
print("---------------max----------")
print(group_df["Age"].max())
print("-------------------------")
print(group_df["Age"].count())
```
